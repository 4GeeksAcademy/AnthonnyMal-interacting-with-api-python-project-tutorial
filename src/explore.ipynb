# Proyecto de explorar spotify

```python
import os
import requests
import pandas as pd
import matplotlib.pyplot as plt
from dotenv import load_dotenv
import spotipy
from spotipy.oauth2 import SpotifyClientCredentials
import seaborn as sns
import numpy as np

# load the .env file variables
load_dotenv()

```

Ya hemos importado librerias 

procedemos a sacar y usar nuestras credenciales con spotify

```python
client_id = os.environ.get("CLIENT_ID")
client_secret = os.environ.get("CLIENT_SECRET")
auth_manager = SpotifyClientCredentials(client_id=client_id, client_secret=client_secret)
spotify = spotipy.Spotify(auth_manager=auth_manager)
```

Seleccionamos al artista y sacamos su ID, y convertimos a DataFrame.

```python
artist_id ="34OLInsXImlQpjzmQw9Wd3"
top_music = spotify.artist_top_tracks(artist_id)["tracks"][:10]
df = pd.DataFrame(top_music)
df
```

Ahora mostraremos el top 3 segun la popularidad

```python
df.sort_values(by='popularity', ascending=False, inplace=True)
df['name'].head(3)
```

Ya podemos ver las 3 canciones con mas popularity.

Usaremos un scatter plot para verificar si el tiempo de una cancion afecta su popularidad

```python
df['duration_seg'] = df['duration_ms'] // 1000
df = df.sort_values('duration_seg')

start = (df['duration_seg'].min() // 30) * 30
end   = ((df['duration_seg'].max() + 29) // 30) * 30  
ticks_seg = list(range(start, end + 1, 30))
labels = [f"{s//60}:{int(s%60):02d}" for s in ticks_seg]
```

```python
sns.scatterplot(x="duration_seg", y="popularity", data=df)
plt.title("Scatter Plot de popularidad según duracion")
plt.xlabel("Duracion de cancion (min)")
plt.ylabel("Popularidad")
plt.xticks(ticks_seg, labels)
plt.show()
```

#### Conclusiones 

Este artista aunque su cancion mas corta es una de las menos pupulares (entre este top), tambien podemos ver que la cancion mas popular de este top es la segunda mas corta. 

Sus 2 canciones mas largas estan en el top 3 de popularidad. 

Podemos asumir que a este artista el tiempo de sus canciones no influye en la popularidad que puedan tener cada una de ellas.
